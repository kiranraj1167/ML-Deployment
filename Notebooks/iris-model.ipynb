```python
import pandas as pd
import pickle
from sklearn.linear_model import LogisticRegression

data=pd.read_csv(r"C:\Users\Kiran\OneDrive\Documents\Downloads\iris_data.csv")
data.shape
```

```python
x=data.iloc[:,:-1].values
y=data.iloc[:,-1].values
```

```python
lr=LogisticRegression(random_state=0)
model=lr.fit(x,y)
```

```python
model.predict([])
```

```python
with open('iris-model.pkl','wb') as f:
    pickle.dump(model, f)
```
